```python
plt.figure(figsize=(6, 4))
sns.histplot(df['Longitude'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Longitude')
plt.show()
```

Above plot shows the distribution of the 'Longitude' feature using a histogram with 20 bins and a KDE (Kernel Density Estimate) overlay. The histogram helps visualize the frequency distribution, while the KDE gives a smoother representation of the data's distribution.

```python
sns.pairplot(df[['Body', 'Sweetness', 'Smoky', 'Longitude']])
plt.suptitle('Pairplot of Selected Features', y=1.02)
plt.show()
```

Above pairplot visualizes the relationships between 'Body', 'Sweetness', 'Smoky', and 'Longitude' through scatter plots and histograms. It allows you to identify potential correlations and distributions of these features in a single view.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='Distillery', y='Longitude', data=df, palette='Set2')
plt.xticks(rotation=90)
plt.title('Longitude by Distillery')
plt.show()
```

Above boxplot shows the distribution of 'Longitude' across different distilleries, highlighting the median, quartiles, and any outliers. It provides insights into the spread of longitude values for each distillery category.

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Features')
plt.show()
```

Above heatmap displays the correlations between the numerical features in the dataset, with color intensity indicating the strength of the relationship. It helps to identify patterns and dependencies between the variables.

```python
plt.figure(figsize=(6, 4))
sns.swarmplot(x='Body', y='Longitude', data=df, palette='Set1')
plt.title('Longitude by Body')
plt.show()
```

The swarmplot shows the distribution of Longitude values across different Body ratings, with each point representing a data point. The color palette 'Set1' is used to differentiate between the various categories of Body.

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='Distillery', data=df, palette='Set3')
plt.xticks(rotation=90)
plt.title('Count of Distilleries')
plt.show()
```

The countplot displays the number of occurrences for each distillery in the dataset, with the x-axis representing different distilleries. The color palette 'Set3' is used to distinguish between the categories, and the x-axis labels are rotated for better readability.

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(x='Body', y='Longitude', data=df, color='purple')
plt.title('Longitude vs Body')
plt.show()
```

The scatterplot shows the relationship between the 'Body' and 'Longitude' columns, with each data point represented by a purple marker. It helps visualize any potential correlation or pattern between the two variables.

```python
plt.figure(figsize=(8, 6))
sns.violinplot(x='Sweetness', y='Longitude', data=df, palette='muted')
plt.title('Longitude by Sweetness')
plt.show()
```

The violin plot displays the distribution of 'Longitude' for each level of 'Sweetness', highlighting the density and range. It also shows the median and interquartile range of 'Longitude' for each sweetness level.

```python
plt.figure(figsize=(6, 4))
sns.barplot(x='Smoky', y='Longitude', data=df, palette='Blues')
plt.title('Longitude by Smoky')
plt.show()
```

The bar plot illustrates the average 'Longitude' for each level of 'Smoky'. It provides insights into how 'Smoky' influences the 'Longitude' values across different categories.

```python
plt.figure(figsize=(6, 4))
sns.kdeplot(df['Longitude'], shade=True, color='green')
plt.title('KDE of Longitude')
plt.show()
```

The KDE plot shows the estimated probability density function of 'Longitude'. It highlights the distribution of 'Longitude' values, emphasizing areas with higher concentrations.

```python
plt.figure(figsize=(6, 4))
sns.stripplot(x='Tobacco', y='Longitude', data=df, jitter=True, palette='coolwarm')
plt.title('Longitude by Tobacco')
plt.show()
```

The strip plot visualizes the distribution of 'Longitude' values across different 'Tobacco' categories. Jittering helps in spreading the data points for better visibility, showing the spread of values within each category.

```python
g = sns.FacetGrid(df, col="Distillery", col_wrap=4, height=3)
g.map(sns.histplot, 'Longitude', kde=True)
plt.suptitle('Longitude Distribution by Distillery', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.heatmap(df[['Body', 'Sweetness', 'Smoky', 'Medicinal', 'Longitude']].corr(), annot=True, cmap='YlGnBu')
plt.title('Correlation Heatmap of Selected Features')
plt.show()
```

The heatmap displays the correlation between selected features like 'Body', 'Sweetness', 'Smoky', 'Medicinal', and 'Longitude'. The annotated values help in identifying strong and weak correlations between these variables.

```python
plt.figure(figsize=(6, 4))
sns.lineplot(x='Latitude', y='Longitude', data=df, marker='o')
plt.title('Longitude vs Latitude')
plt.show()
```

The line plot visualizes the relationship between 'Longitude' and 'Latitude' across the dataset. It shows how changes in latitude correlate with variations in longitude, marked by data points.
